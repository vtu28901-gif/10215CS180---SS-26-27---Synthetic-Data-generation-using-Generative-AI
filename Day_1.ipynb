# **Part F — What should our synthetic data preserve?**

```python
baseline = df.sample(
    n=len(df),
    replace=True,                         #bootstrap resampling
    random_state=42
).reset_index(drop=True)
```

```python
baseline.shape
```

```python
numeric_cols = [
    "glucose_mg_dl",        #comparing numerical
    "heart_rate_bpm",
    "hrv_ms",
    "spo2_percent",
    "sleep_duration_hours",
    "steps"
]

comparison = pd.DataFrame({
    "real_mean": df[numeric_cols].mean(),
    "baseline_mean": baseline[numeric_cols].mean(),
    "real_std": df[numeric_cols].std(),
    "baseline_std": baseline[numeric_cols].std()
}).round(2)

comparison
```

```python
activity_compare = pd.DataFrame({
    "real_%": df["activity_level"].value_counts(normalize=True) * 100, #Compare categorical
    "baseline_%": baseline["activity_level"].value_counts(normalize=True) * 100
}).round(2)

activity_compare
```

```python
real_corr = df[numeric_cols].corr()
baseline_corr = baseline[numeric_cols].corr()

corr_delta = baseline_corr - real_corr    #Compare relationships between baseline and real

corr_delta.round(2)
```
