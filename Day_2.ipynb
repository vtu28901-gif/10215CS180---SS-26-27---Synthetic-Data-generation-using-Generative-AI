# **DAY 3**

```python
# Day 3 — Fidelity: compare distribution quantiles

numeric_cols = [
    "glucose_mg_dl",
    "heart_rate_bpm",
    "hrv_ms",
    "spo2_percent",
    "sleep_duration_hours",
    "steps"
]

quantiles = [0.05, 0.25, 0.50, 0.75, 0.95]

for col in numeric_cols:
    print(f"\n{'='*60}")
    print(col)

    real_q = ctgan_data[col].quantile(quantiles)
    synthetic_q = synthetic_data[col].quantile(quantiles)

    comparison = pd.DataFrame({
        "real": real_q,
        "synthetic": synthetic_q,
        "difference": synthetic_q - real_q
    }).round(2)

    print(comparison)
```

```python
from sklearn.model_selection import train_test_split    #Prepare the data
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

target = "activity_level"

features = [
    "glucose_mg_dl",
    "heart_rate_bpm",
    "hrv_ms",
    "spo2_percent",
    "sleep_duration_hours",
    "steps",
    "device_type"
]

# Real data
real_model_data = ctgan_data[features + [target]].copy()

# Synthetic data
synthetic_model_data = synthetic_data[features + [target]].copy()

X_real = real_model_data[features]
y_real = real_model_data[target]

X_synthetic = synthetic_model_data[features]
y_synthetic = synthetic_model_data[target]
```

```python
X_real_train, X_real_test, y_real_train, y_real_test = train_test_split( #Split the real data
    X_real,
    y_real,
    test_size=0.20,
    random_state=42,
    stratify=y_real
)

print("Real training rows:", len(X_real_train))
print("Real test rows:", len(X_real_test))
```

```python
numeric_features = [
    "glucose_mg_dl",
    "heart_rate_bpm",
    "hrv_ms",
    "spo2_percent",
    "sleep_duration_hours",
    "steps"
]

categorical_features = [
    "device_type"
]
numeric_pipeline = Pipeline([          #Build the model
    ("imputer", SimpleImputer(strategy="median"))
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("num", numeric_pipeline, numeric_features),
    ("cat", categorical_pipeline, categorical_features)
])

model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=100,
        random_state=42
    ))
])
```

```python
# Train ONLY on synthetic data
model.fit(X_synthetic, y_synthetic)

# Test ONLY on unseen real data
y_pred = model.predict(X_real_test)

tstr_accuracy = accuracy_score(y_real_test, y_pred)

print("TSTR Accuracy:", round(tstr_accuracy, 3))

print("\nClassification Report:")
print(classification_report(y_real_test, y_pred))
```

```python
real_model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=100,
        random_state=42
    ))
])

# Train on real training data
real_model.fit(X_real_train, y_real_train)

# Test on the same real test data
y_real_pred = real_model.predict(X_real_test)

real_accuracy = accuracy_score(y_real_test, y_real_pred)

print("Real → Real Accuracy:", round(real_accuracy, 3))
print("Synthetic → Real Accuracy:", round(tstr_accuracy, 3))
```

```python
# Check whether synthetic rows exactly match real rows i.e, Privacy Check 1 — Exact duplicate rows

real_check = ctgan_data.reset_index(drop=True)
synthetic_check = synthetic_data.reset_index(drop=True)

# Convert rows to tuples so we can compare them
real_rows = set(real_check.astype(str).apply(tuple, axis=1))

exact_matches = synthetic_check.astype(str).apply(
    tuple, axis=1
).isin(real_rows)

print("Synthetic rows:", len(synthetic_check))
print("Exact matches:", exact_matches.sum())
print(
    "Exact match percentage:",
    round(exact_matches.mean() * 100, 2),
    "%"
)
```

```python
from sklearn.impute import SimpleImputer          #Privacy Check 2 — Check for very similar records
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import NearestNeighbors

privacy_numeric = [
    "glucose_mg_dl",
    "heart_rate_bpm",
    "hrv_ms",
    "spo2_percent",
    "sleep_duration_hours",
    "steps"
]

real_numeric = ctgan_data[privacy_numeric]
synthetic_numeric = synthetic_data[privacy_numeric]

# Fill missing values using real-data medians
imputer = SimpleImputer(strategy="median")

real_imputed = imputer.fit_transform(real_numeric)
synthetic_imputed = imputer.transform(synthetic_numeric)

# Put variables on comparable scales
scaler = StandardScaler()

real_scaled = scaler.fit_transform(real_imputed)
synthetic_scaled = scaler.transform(synthetic_imputed)

# Find nearest real record for every synthetic record
nn = NearestNeighbors(n_neighbors=1)
nn.fit(real_scaled)

distances, indices = nn.kneighbors(synthetic_scaled)

nearest_distances = distances[:, 0]

print("Minimum distance:", round(nearest_distances.min(), 3))
print("Median distance:", round(float(__import__("numpy").median(nearest_distances)), 3))
print("5th percentile:", round(float(__import__("numpy").percentile(nearest_distances, 5)), 3))
print("95th percentile:", round(float(__import__("numpy").percentile(nearest_distances, 95)), 3))
```
